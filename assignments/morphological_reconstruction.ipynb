# Remove complex background using morphological reconstruction

- A nice example of clever application of morphology is illustrated in the [Gongalez18] book.
- There, we have an image of a calculator and we want to remove the background consisting of the buttons and the reflections.
- As an assignment, we're going to implement the example in Python.

**Morphological approach to background removal**
1. First we're going to remove most of the background using top-hat by reconstruction (tasks 2 and 3).
2. As a next step we're going to remove residual artifacts, mostly vertical reflections.
3. This will unfortunately erase some of the characters as well, so we will bring them back using dilation.
4. As the final step we're goind to use the dilated image as a starting point a reconstruct it w.r.t to the top-hat image from step 1.
